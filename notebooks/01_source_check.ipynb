```python
from pathlib import Path
import re
import docx

ROOT = Path.cwd().parent          # notebook "notebooks/" içinde çalışır, bir üstü repo kökü
WORD_DIR = ROOT / "data" / "legal" / "word"

# PDF sürümünde sayılan madde başlıkları (2/11/2026 karşılaştırması)
EXPECTED = {
    "01_kanun_2918": 193,
    "02_trafik_yonetmeligi": 222,
    "03_meb_surucu_kursu": 74,
    "04_surucu_saglik": 14,
    "05_arac_muayene": 30,
    "06_elektrikli_skuter": 34,
}

HEADER = re.compile(r"^\s*(?:Ek |Geçici |EK |GEÇİCİ )?(?:Madde|MADDE)\s+\d+(?:/[A-ZÇ])?\s*[–\-—]")

for name, expected in EXPECTED.items():
    d = docx.Document(WORD_DIR / f"{name}.docx")
    texts = [p.text for p in d.paragraphs]
    headers = sum(1 for t in texts if HEADER.match(t))
    turkish_ok = all(c in "".join(texts) for c in "şğİ")
    superscripts = sum(1 for p in d.paragraphs for r in p.runs if r.font.superscript)
    status = "OK" if headers == expected and turkish_ok else "KONTROL ET"
    print(f"{name:25} başlık {headers:3}/{expected:3}  Türkçe:{turkish_ok}  üst simge:{superscripts:3}  {status}")
```
